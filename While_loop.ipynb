**Basic While Loops Problems**
1. Write a program that prints all even numbers from 2 to 20 using a while loop.
2. Write a program that prints all odd numbers between 1 and 15. (assignment)
3. Create a countdown program from 10 to 1, then print "Blast off!".
4. Write a program to calculate the sum of numbers from 1 to 100 using a while loop.
5. Write a program that asks the user to enter a number and prints its multiplication table from 1 to 12.
6. Write a program that keeps printing "Hello" 5 times using a while loop. Assignment
7. Write a program that asks the user for a password repeatedly until the correct password is entered.
8. Write a program that keeps asking the user for numbers and stops when the user enters 0.
9. Write a program that counts how many digits are in a given number.

```python
while True:
    print("\n===== WHILE LOOP PRACTICE MENU =====")
    print("1. Even numbers from 2 to 20")
    print("2. Odd numbers from 1 to 15")
    print("3. Countdown from 10 to 1")
    print("4. Sum of numbers from 1 to 100")
    print("5. Multiplication table 1 to 12")
    print("6. Print Hello 5 times")
    print("7. Password checker")
    print("8. Sum numbers until 0 is entered")
    print("9. Count digits in a number")
    print("0. Quit/n/n")

    choice = input("Enter your choice: ").strip()

    if choice == "1":
        num = 2
        while num <= 20:
            print(num)
            num += 2

    elif choice == "2":
        num = 1
        while num <= 15:
            print(num)
            num += 2

    elif choice == "3":
        count = 10
        while count >= 1:
            print(count)
            count -= 1
        print("Blast off!")

    elif choice == "4":
        num = 1
        total = 0
        while num <= 100:
            total += num
            num += 1
        print("Sum:", total)

    elif choice == "5":
        try:
            number = int(input("Enter a number: "))
            i = 1
            while i <= 12:
                print(number, "x", i, "=", number * i)
                i += 1
        except ValueError:
            print("Please enter a valid number")

    elif choice == "6":
        count = 1
        while count <= 5:
            print("Hello")
            count += 1

    elif choice == "7":
        password = "python123"
        user_input = ""
        while user_input != password:
            user_input = input("Enter password: ")
        print("Access granted")

    elif choice == "8":
        total = 0
        num = -1
        while num != 0:
            try:
                num = int(input("Enter a number, 0 to stop: "))
                total += num
            except ValueError:
                print("Please enter a valid number")
        print("Total sum:", total)

    elif choice == "9":
        try:
            number = abs(int(input("Enter a number: ")))
            count = 0
            if number == 0:
                count = 1
            else:
                while number > 0:
                    number = number // 10
                    count += 1
            print("Number of digits:", count)
        except ValueError:
            print("Please enter a valid number")

    elif choice == "0":
        print("Goodbye!")
        break

    else:
        print("Invalid choice. Try again.")
```
