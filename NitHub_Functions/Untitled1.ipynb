#Problems
1. Write a function that takes the length and width of a rectangle and returns its total perimeter.
2. Write a function that takes a person's age. It should return 10 (dollars) if the age is 12 or under, and 15 (dollars) if they are older.
3. Write a function is_valid_username(username) that takes a string. It should return *True* if the username is at least 6 characters long, and *False* if it is shorter.
4. Write a function that takes an integer. It should return the string "Even" if the number is even, and "Odd" if the number is odd.
5. Write a function is_strong_password(password) that takes a string. It should return True ONLY if the password meets two rules: it must be at least 8 characters long AND it must contain at least one exclamation point (!). Otherwise, return False.
This should be submitted on the worksheet labelled Assignment (17/06/26)

```python
#Case 1
def rectangle_perimeter(length, width):
    sum_of_sides = length + width
    perimeter = sum_of_sides * 2
    return perimeter

while True:
    length = int(input('What is the length of your rectangle? '))
    width = int(input('What is the width of your rectangle? '))

    perimeter = rectangle_perimeter(length, width)
    print(f'Perimeter is {perimeter} units')  # note: perimeter is in units, not sq units

    again = input('Do you want to calculate another one? (y/n): ').lower()
    if again == 'n' or again == 'q':
        print('Goodbye!')
        break
```

```python
#Case 2
def ticket_price(age):
    if age < 0:
        print('Please enter a valid age')
        return None
    elif age <= 12:
        price = 10
    else:
        price = 15

    return price

while True:
    age = int(input('Please enter your age: '))
    price = ticket_price(age)

    if price is not None:
        print(f'Your ticket price is ${price}')

    again = input('Do you want to check another ticket? (y/n): ').lower()
    if again == 'n':
        print('Goodbye!')
        break
```

```python

#Case 3
def is_valid_username(username):
    length_of_username = len(username)
    if length_of_username >= 6:
        return True
    else:
        return False

while True:
    input_username = str(input('Please input your username: '))
    valid = is_valid_username(input_username)

    if valid:
        print('Username is valid')
    else:
        print('Username is too short. Must be at least 6 characters')

    again = input('Do you want to try another username? (y/n): ').lower()
    if again == 'n':
        print('Goodbye!')
        break
```

```python
#Case 4
def even_or_odd(number):
    remainder = number % 2
    if remainder == 0:
        result = "Even"
    else:
        result = "Odd"

    return result

while True:
    num = int(input('Please input number: '))
    result = even_or_odd(num)
    print(f'The number is {result}')

    again = input('Do you want to check another number? (y/n): ').lower()
    if again == 'n':
        print('Goodbye!')
        break
```

```python
#Case 5
def is_strong_password(password):
    long_enough = len(password) >= 8
    has_exclamation = "!" in password
    if long_enough and has_exclamation:
        return True
    else:
        return False

while True:
    password = str(input('Please create a password: '))
    strong = is_strong_password(password)

    if strong:
        print('Password is strong')
    else:
        print('Password is weak. It needs at least 8 characters and a "!"')

    again = input('Do you want to try another password? (y/n): ').lower()
    if again == 'n':
        print('Goodbye!')
        break
```
