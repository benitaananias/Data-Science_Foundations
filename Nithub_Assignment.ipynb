Even numbers

```python
for n in range(1, 21):
    if n % 2 == 0:
        print(n)
```

Print numbers from 1-5

```python
for n in range(1, 6):
    if n == 3:
        continue
    print(n)
```

Vowel count

```python
word = input("Enter a word: ").lower()
vowels = "aeiou"
count = 0
for w in word:
    if w in vowels:
        count += 1
print(f"Vowels: {count}")
```
