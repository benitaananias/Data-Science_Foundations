A voter eligiblity system to asses the eligiblity of voters according to set conditions and user inputs.

```python

print("WELCOME TO THE VOTERS ELIGIBILITY PAGE")
print("Kindly confirm if you are of age and the right nationality\n")

while True:
    try:
        voters_age = int(input("Kindly enter your age in digits: "))

        if voters_age < 18 or voters_age > 85:
            print("You are not eligible to Vote: Age not within 18-85")

        else:
            voters_nationality = input("Are you a Nigerian? yes/no: ").strip().lower()

            if voters_nationality == "yes":
                print("You are eligible to Vote")
            else:
                print("You are not eligible to Vote: You must be a Nigerian citizen")

        again = input("\nCheck another person? yes/no: ").strip().lower()
        if again != "yes":
            print("Goodbye!")
            break

    except ValueError:
        print("Please enter a valid number for age")
```
