1. Create a NumPy array of 50 random integers between 1 and 100.
    * Compute:
      - Mean
      - Median
      - Standard deviation
    * Find:
        - Maximum value
        - Minimum value

```python
import numpy as np

# Create array of 50 random integers between 1 and 100
arr = np.random.randint(1, 101, 50)

# Compute mean, median, standard deviation
mean = np.mean(arr)
median = np.median(arr)
standad_deviation = np.std(arr)

# Find max and min
maximum = np.max(arr)
minimum = np.min(arr)

# Print results
print("Array:", arr)
print("Mean:", mean)
print("Median:", median)
print("Standard Deviation:", standad_deviation)
print("Maximum value:", maximum)
print("Minimum value:", minimum)
```

2. Using the array created:
    * Extract all values:
        - Greater than 60
        - Less than 30

```python
print("Array:", arr)

greater_than_60 = arr[arr > 60]
less_than_30 = arr[arr < 30]

print("Values greater than 60:", greater_than_60)
print("Values less than 30:", less_than_30)
```

3. Using the employee dataset I sent in one of our previous classes:
   * Display the first 5 rows
   * Display summary statistics
   * Identify the job title with the highest `salary_year_avg`
   * Count the number of unique job titles

```python
import pandas as pd

df = pd.read_csv('employee_dataset.csv')
print("First five rows", df.head())

print("Summay statistics", df.describe())

print("Highest salary per year", df.loc[df['salary_year_avg'].idxmax(), 'job_title'])

print("Number of unique job titles", df['job_title'].nunique())
```

4. Using the same dataset:
   * Filter `salary_year_avg` above 70000
   * Sort job title by `salary_year_avg` (descending order)

```python
# filter salary_year_avg above 70000
high_salary = df[df['salary_year_avg'] > 70000]
print(high_salary)

# sort job titles by salary_year_avg
sorted_jobs = df.sort_values(by='salary_year_avg', ascending=False)
print(sorted_jobs[['job_title', 'salary_year_avg']])
```

5. Using the same data
    * Plot a bar chart of job title vs `salary_year_avg`
    * Plot a line graph of `salary_year_avg`
    * Add:
      - Title
      - Axis Labels

```python
# bar chart
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.bar(df['job_title'], df['salary_year_avg'])

plt.title("Job Title vs Average Salary")
plt.xlabel("Job Title")
plt.ylabel("Salary Year Average")

plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()

# line graph
plt.figure(figsize=(10,5))

plt.plot(df['salary_year_avg'])

plt.title("Salary Year Average Trend")
plt.xlabel("Index")
plt.ylabel("Salary Year Average")

plt.show()
```
