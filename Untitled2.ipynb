A grading system that takes user inputted scores and gives letter coded grades accordingly:

```python

print("""
GRADING SYSTEM
Students are to fill in their score out of 100 to see what grade they attained.
Type -1 to quit
""")

while True:
    try:
        score = int(input("Enter your score: "))
    except ValueError:
        print("Please enter a number")
        continue

    if score == -1:
        print("Goodbye!")
        break
    elif score > 100 or score < 0:
        print("Invalid score")
    elif score >= 70:
        print("Grade: A")
    elif score >= 60:
        print("Grade: B")
    elif score >= 50:
        print("Grade: C")
    elif score >= 45:
        print("Grade: D")
    elif score >= 39:
        print("Grade: E")
    else:
        print("Grade: F")  # changed FAIL to F to match the pattern
```
